All visits with ended , confirmed

```python
steps = ['start','step_1','step_2','step_3','confirm']
```

```python
def error_by_step(df):
    
    all_visits = df['visit_id'].nunique()

    for i in range(len(steps) -1): ## -1 , not for confirmed steps
        step_current = steps[i]
        step_next = steps[i+1]

        visit_step_current = set(df[df['process_step'] == step_current]['visit_id'])
        visit_step_next = set(df[df['process_step'] == step_next]['visit_id'])

        error = visit_step_current - visit_step_next

        percentage_error = (len(error) / all_visits)*100

        print(f'Error on {step_current} to {step_next} is: {percentage_error:.2f}%')
    
```

```python
# A/B test:
# Determine if the completion rate ( confirm step) is higher in Test Group than in the Control Group.
```

```python
#Hypothesis

#H0: mu_confirm_test  <= mu_confirm_control
#H1: mu_confirm_test > mu_confirm_control

#significance level = 0.05
```

```python
import scipy.stats as st
```

```python
from statsmodels.stats.proportion import proportions_ztest
```

```python
succes = [confirm_test,confirm_control]
succes
```

```python
sample = [visits_test,visits_control]
sample
```

```python
stat, p_value = proportions_ztest(succes,sample, alternative='larger')
stat, p_value
```

```python
p_value1 = 6.247505366322125e-116
```

```python
alpha = 0.05
if p_value < alpha:
    print('Reject null hypothesis.')
else:
    print('Do not reject null hypothesis.')
```
